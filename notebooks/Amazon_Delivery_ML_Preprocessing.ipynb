```python
# IQR Based Outlier Detection

outlier_columns = [
    "Agent_Age",
    "Agent_Rating",
    "Delivery_Time",
    "Pickup_Delay_Minutes"
]

outlier_results = []

for col in outlier_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    outlier_results.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Limit": lower_limit,
        "Upper_Limit": upper_limit,
        "Outlier_Count": len(outliers),
        "Outlier_Percentage": round(
            len(outliers) / len(df) * 100, 2
        )
    })

outlier_report = pd.DataFrame(outlier_results)

display(outlier_report)
```

```python
# Separate features and target

X = ml_df.drop(columns=["Delivery_Time"])
y = ml_df["Delivery_Time"]

print("X Shape:", X.shape)
print("y Shape:", y.shape)
```

```python
# Separate numerical and categorical features

numerical_features = X.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object", "str", "category"]
).columns.tolist()

print("Numerical Features:")
print(numerical_features)

print("\nTotal Numerical Features:", len(numerical_features))

print("\nCategorical Features:")
print(categorical_features)

print("\nTotal Categorical Features:", len(categorical_features))
```

```python
# Create a separate ML working dataset

ml_df = df.copy()

# Columns to exclude
drop_columns = [
    "Order_ID",
    "Order_Year",
    "Order_Date",
    "Order_Time",
    "Pickup_Time",
    "Pickup_Hour",
    "Pickup_Minute",
    "Pickup_Delay_Minutes",
    "Order_Time_Parsed",
    "Pickup_Time_Parsed"
]

# Drop unnecessary columns
ml_df.drop(columns=drop_columns, inplace=True, errors="ignore")

print("Original Dataset Shape:", df.shape)
print("ML Working Dataset Shape:", ml_df.shape)

print("\nRemaining Columns:")
print(ml_df.columns.tolist())

print("\nMissing Values:")
print(ml_df.isnull().sum())
```

```python
from sklearn.model_selection import train_test_split

# Split features and target
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training Data Shape:", X_train.shape)
print("Testing Data Shape:", X_test.shape)

print("\nTraining Target Shape:", y_train.shape)
print("Testing Target Shape:", y_test.shape)

print("\nTraining Missing Values:", X_train.isnull().sum().sum())
print("Testing Missing Values:", X_test.isnull().sum().sum())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Numerical preprocessing
numerical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical preprocessing
categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="constant", fill_value="Unknown")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine both pipelines
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numerical_pipeline, numerical_features),
        ("cat", categorical_pipeline, categorical_features)
    ]
)

print("Preprocessing Pipeline Created Successfully!")
```

```python
# Fit on training data and transform
X_train_processed = preprocessor.fit_transform(X_train)

# Transform testing data
X_test_processed = preprocessor.transform(X_test)

print("Training Processed Shape:", X_train_processed.shape)
print("Testing Processed Shape:", X_test_processed.shape)

print("\nTraining Missing Values:", 
      X_train_processed.sum() if False else "Checking...")

print("\nPreprocessing Completed Successfully!")
```

```python
import numpy as np
from scipy import sparse

# Check missing values in training data
if sparse.issparse(X_train_processed):
    train_missing = np.isnan(X_train_processed.data).sum()
else:
    train_missing = np.isnan(X_train_processed).sum()

# Check missing values in testing data
if sparse.issparse(X_test_processed):
    test_missing = np.isnan(X_test_processed.data).sum()
else:
    test_missing = np.isnan(X_test_processed).sum()

print("Training Missing Values:", train_missing)
print("Testing Missing Values:", test_missing)
```

```python
feature_names = preprocessor.get_feature_names_out()

print("Total Processed Features:", len(feature_names))

print("\nProcessed Feature Names:")
for feature in feature_names:
    print(feature)
```

```python
import pandas as pd
from scipy import sparse

# Convert processed arrays into DataFrames
X_train_df = pd.DataFrame(
    X_train_processed.toarray() if sparse.issparse(X_train_processed) else X_train_processed,
    columns=feature_names
)

X_test_df = pd.DataFrame(
    X_test_processed.toarray() if sparse.issparse(X_test_processed) else X_test_processed,
    columns=feature_names
)

# Add target column
X_train_df["Delivery_Time"] = y_train.reset_index(drop=True)
X_test_df["Delivery_Time"] = y_test.reset_index(drop=True)

print("Final Training Dataset:", X_train_df.shape)
print("Final Testing Dataset:", X_test_df.shape)

print("\nTraining Dataset Preview:")
display(X_train_df.head())

print("\nTesting Dataset Preview:")
display(X_test_df.head())
```
