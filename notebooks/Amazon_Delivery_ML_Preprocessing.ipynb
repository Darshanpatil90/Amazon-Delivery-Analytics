```python
# IQR Based Outlier Detection

outlier_columns = [
    "Agent_Age",
    "Agent_Rating",
    "Delivery_Time",
    "Pickup_Delay_Minutes"
]

outlier_results = []

for col in outlier_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    outlier_results.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Limit": lower_limit,
        "Upper_Limit": upper_limit,
        "Outlier_Count": len(outliers),
        "Outlier_Percentage": round(
            len(outliers) / len(df) * 100, 2
        )
    })

outlier_report = pd.DataFrame(outlier_results)

display(outlier_report)
```

```python
# Validate Potential Outliers

print("1. DELIVERY TIME ABOVE IQR UPPER LIMIT")
print("-" * 50)

delivery_outliers = df[df["Delivery_Time"] > 265]

print("Outlier Count:", len(delivery_outliers))
print("\nDelivery Time Frequency:")
display(delivery_outliers["Delivery_Time"].value_counts().sort_index())


print("\n2. AGENT RATING DISTRIBUTION")
print("-" * 50)

print("Minimum Rating:", df["Agent_Rating"].min())
print("Maximum Rating:", df["Agent_Rating"].max())

display(
    df["Agent_Rating"]
    .value_counts()
    .sort_index()
    .to_frame("Frequency")
)


print("\n3. DELIVERY TIME PERCENTILES")
print("-" * 50)

display(
    df["Delivery_Time"]
    .quantile([0.90, 0.95, 0.99, 0.995, 1.00])
    .to_frame("Delivery_Time")
)
```
