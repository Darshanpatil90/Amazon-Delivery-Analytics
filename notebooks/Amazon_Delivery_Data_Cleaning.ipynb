```python
import datetime

def time_to_minutes(t):
    if pd.isna(t):
        return np.nan

    return (
        t.hour * 60 +
        t.minute +
        t.second / 60
    )

order_minutes = df_cleaned['Order_Time'].apply(time_to_minutes)
pickup_minutes = df_cleaned['Pickup_Time'].apply(time_to_minutes)

time_difference = pickup_minutes - order_minutes

# Handle midnight crossing
time_difference = time_difference % 1440

print("Average Pickup Delay:", time_difference.mean())
print("Median Pickup Delay:", time_difference.median())
print("Minimum Delay:", time_difference.min())
print("Maximum Delay:", time_difference.max())
```

```python
# Extract date-related features

df_cleaned['Order_Year'] = df_cleaned['Order_Date'].dt.year
df_cleaned['Order_Month'] = df_cleaned['Order_Date'].dt.month
df_cleaned['Order_Day'] = df_cleaned['Order_Date'].dt.day
df_cleaned['Order_DayOfWeek'] = df_cleaned['Order_Date'].dt.dayofweek

print(df_cleaned[
    ['Order_Date', 'Order_Year', 'Order_Month',
     'Order_Day', 'Order_DayOfWeek']
].head())
```

```python
# Extract Order Hour and Minute

df_cleaned['Order_Hour'] = df_cleaned['Order_Time'].apply(
    lambda t: t.hour if pd.notna(t) else np.nan
)

df_cleaned['Order_Minute'] = df_cleaned['Order_Time'].apply(
    lambda t: t.minute if pd.notna(t) else np.nan
)

print(df_cleaned[
    ['Order_Time', 'Order_Hour', 'Order_Minute']
].head())

print("\nMissing Order_Hour:", df_cleaned['Order_Hour'].isnull().sum())
```

```python
# Extract Pickup Hour and Minute

df_cleaned['Pickup_Hour'] = df_cleaned['Pickup_Time'].apply(
    lambda t: t.hour if pd.notna(t) else np.nan
)

df_cleaned['Pickup_Minute'] = df_cleaned['Pickup_Time'].apply(
    lambda t: t.minute if pd.notna(t) else np.nan
)

print(df_cleaned[
    ['Pickup_Time', 'Pickup_Hour', 'Pickup_Minute']
].head())

print("\nMissing Pickup_Hour:",
      df_cleaned['Pickup_Hour'].isnull().sum())
```

```python
def time_to_minutes(t):
    if pd.isna(t):
        return np.nan
    return t.hour * 60 + t.minute

order_minutes = df_cleaned['Order_Time'].apply(time_to_minutes)
pickup_minutes = df_cleaned['Pickup_Time'].apply(time_to_minutes)

df_cleaned['Pickup_Delay_Minutes'] = (
    pickup_minutes - order_minutes
) % 1440

print(df_cleaned[
    ['Order_Time', 'Pickup_Time', 'Pickup_Delay_Minutes']
].head())

print("\nAverage Pickup Delay:",
      df_cleaned['Pickup_Delay_Minutes'].mean())

print("Missing Pickup Delay:",
      df_cleaned['Pickup_Delay_Minutes'].isnull().sum())
```

```python
def get_time_period(hour):
    if pd.isna(hour):
        return np.nan
    elif 5 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Afternoon'
    elif 17 <= hour < 21:
        return 'Evening'
    else:
        return 'Night'


df_cleaned['Order_Time_Period'] = df_cleaned['Order_Hour'].apply(get_time_period)

print(df_cleaned[
    ['Order_Time', 'Order_Hour', 'Order_Time_Period']
].head())

print("\nTime Period Distribution:")
print(df_cleaned['Order_Time_Period'].value_counts(dropna=False))
```
